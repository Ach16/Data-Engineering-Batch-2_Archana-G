```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("SparkSQL Lab").getOrCreate()
```

CREATING DATAFRAMES

```python
employees_data = [
    (1, "Rahul", "IT", 70000, "Hyderabad"),
    (2, "Sneha", "HR", 60000, "Bangalore"),
    (3, "Arjun", "IT", 75000, "Chennai"),
    (4, "Priya", "Finance", 80000, "Hyderabad"),
    (5, "Karan", "IT", 50000, "Mumbai"),
    (6, "Amit", "HR", 58000, "Delhi"),
    (7, "Meera", "Finance", 82000, "Bangalore")
]

employees_cols = ["emp_id", "name", "department", "salary", "city"]
employees_df = spark.createDataFrame(employees_data, employees_cols)
employees_df.show()
```

```python
departments_data = [
    ("IT", "Technology"),
    ("HR", "People Operations"),
    ("Finance", "Accounts and Finance")
]

departments_cols = ["department", "dept_full_name"]
departments_df = spark.createDataFrame(departments_data, departments_cols)
departments_df.show()
```

```python
sales_data = [
    (101, 1, "Laptop", 1, 75000),
    (102, 2, "Mouse", 3, 500),
    (103, 3, "Keyboard", 2, 1500),
    (104, 1, "Monitor", 1, 12000),
    (105, 4, "Laptop", 1, 75000),
    (106, 3, "Mouse", 2, 500),
    (107, 5, "Keyboard", 1, 1500),
    (108, 1, "Laptop", 1, 75000)
]

sales_cols = ["sale_id", "emp_id", "product", "quantity", "price"]
sales_df = spark.createDataFrame(sales_data, sales_cols)
sales_df.show()
```

CREATING TEMPORARY VIEWS FOR THE  DATAFRAMES (WE CAN ONLY USE THAT IN THE CURRENT SESSION)

```python
employees_df.createOrReplaceTempView("employees")
departments_df.createOrReplaceTempView("departments")
sales_df.createOrReplaceTempView("sales")
```

SOME BASIC SQL OPERATIONS IN PYSPARK

```python
spark.sql("""
CREATE OR REPLACE VIEW high_salary_employees AS
SELECT emp_id,name,department,salary
FROM employee_details
WHERE salary>70000
""")
```

```python
spark.sql("SELECT * FROM high_salary_employees").show()
```

```python
spark.sql("""
SELECT name,city
FROM employees
WHERE department ="IT"
""").show()
```

```python
spark.sql("""
SELECT *
FROM employees
WHERE city In("Hyderabad","Bangalore")
""").show()
```

JOINS

```python
spark.sql("""
SELECT e.emp_id,e.name,e.department,d.dept_full_name
FROM employees e
INNER JOIN departments d
ON e.department = d.department
""").show()
```

```python
spark.sql("""
SELECT e.emp_id,e.name,e.department,d.dept_full_name
FROM employees e
LEFT JOIN departments d
ON e.department = d.department
""").show()
```

AGGREGATION

```python
spark.sql("""
SELECT
COUNT(*) AS total_employees,
SUM(salary) AS total_salary,
AVG(salary) AS avg_salary,
MAX(salary) AS max_salary,
MIN(salary) AS min_salary
FROM employees
""").show()
```

GROUP BY

```python
spark.sql("""
SELECT department,COUNT(*) AS emp_count
FROM employees
GROUP BY department
""").show()
```

```python
spark.sql("""
SELECT department, AVG(salary) AS avg_salary
FROM employees
GROUP BY department
""").show()
```

```python
spark.sql("""
SELECT e.name,SUM(s.quantity*s.price) AS total_sales
FROM employees e
JOIN sales s
ON e.emp_id = s.emp_id
GROUP BY e.name
""").show()
```

HAVING CLAUSE

```python
spark.sql("""
SELECT department,AVG(salary) AS avg_salary
FROM employees
GROUP BY department
HAVING avg_salary>65000
""").show()
```

```python
spark.sql("""
SELECT product,SUM(quantity) AS total_qty
FROM sales
GROUP BY product
HAVING total_qty>2
""").show()
```

ORDER BY

```python
spark.sql("""
SELECT *
FROM employees
ORDER BY salary DESC
""").show()
```

```python
spark.sql("""
SELECT product,SUM(quantity*price) AS revenue
FROM sales
GROUP BY product
ORDER BY revenue DESC
""").show()
```

WINDOW FUNCTIONS

RANK() - ASSIGNS RANK BUT THERE WILL BE GAPS. EG:  
70000 - 1  
70000 - 1  
65000 - 3  
RANK 2 IS MISSING



```python
spark.sql("""
SELECT
emp_id,
name,
department,
salary,
RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS salary_rank
FROM employees
""").show()
```

ROW_NUMBER() - JUST ASSIGNS A ROW NUMBER INSTEAD OF RANKING

```python
spark.sql("""
SELECT
emp_id,
name,
department,
salary,
ROW_NUMBER() OVER (PARTITION BY department ORDER BY salary DESC) AS row_num
FROM employees
""").show()
```

DENSE_RANK() - ASSIGNS RANK WITH NO GAP  
EG:  
70000 - 1  
70000 - 1  
65000 - 2  
RANK 2 IS NOT MISSING


```python
spark.sql("""
SELECT
emp_id,
name,
department,
salary,
DENSE_RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS dense_rank
FROM employees
""").show()
```
