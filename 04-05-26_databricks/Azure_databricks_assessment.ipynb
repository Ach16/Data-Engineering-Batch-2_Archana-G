PART 1 — DATAFRAME (PYSPARK)

```python
#1
from pyspark.sql import SparkSession
spark=SparkSession.builder.getOrCreate()
data = [
(101,"Arjun Reddy","Hyderabad","Cardiology",5000,1),
(102,"Sneha Kapoor","Delhi","Orthopedics",3000,2),
(103,"Rahul Sharma","Mumbai","Dermatology",1500,1),
(104,"Priya Nair","Bangalore","Cardiology",5000,2),
(105,"Vikram Singh","Chennai","Neurology",7000,1),
(106,"Ananya Das","Kolkata","Orthopedics",3000,3),
(107,"Karan Patel","Ahmedabad","Cardiology",5000,1),
(108,"Meera Iyer","Bangalore","Dermatology",1500,2)
]
columns=["visit_id","patient_name","city","department","consultation_fee","tests_count"]
df=spark.createDataFrame(data,columns)
df.show()
```

```python
#2
from pyspark.sql.functions import col
df=df.withColumn("total_cost",col("consultation_fee")+col("tests_count")*500)
df.show()
```

```python
#3
df.filter(col("total_cost")>5000).show()
```

```python
#4
from pyspark.sql.functions import count,sum,avg
df.groupBy("department").agg(count("*").alias("patient_count"),sum("total_cost").alias("total_revenue")).show()
```

```python
#5
df.orderBy(col("total_cost").desc()).show()
```

```python
new_data =[(103,"Sneha Kapoor","Delhi", "Orthopedics",3500,2,4500),(110,"Amit Sharma","Mumbai","Cardiology",5000,1,5500)]
new_df=spark.createDataFrame(new_data,df.schema)
new_df.createOrReplaceTempView("updates_view")

spark.sql("""
MERGE INTO patients_delta AS target
USING updates_view AS source
ON target.visit_id = source.visit_id
WHEN MATCHED THEN
UPDATE SET
target.patient_name=source.patient_name,
target.city=source.city,
target.department=source.department,
target.consultation_fee=source.consultation_fee,
target.tests_count=source.tests_count,
target.total_cost=source.total_cost
WHEN NOT MATCHED THEN INSERT *
""")
spark.sql("SELECT * FROM patients_delta ORDER BY visit_id").show()
```

PART 6 — INCREMENTAL LOAD

```python
%sql
DROP TABLE IF EXISTS patients_target;
CREATE TABLE patients_target
USING DELTA
AS SELECT * FROM VALUES
(115,'Rahul Verma','Hyderabad','Cardiology',5200,2,10400),
(116,'Neha Sharma','Delhi','Dermatology',1800,3,5400),
(117,'Vivek Nair','Kochi','Neurology',7500,1,7500),
(118,'Pooja Kulkarni','Pune','Orthopedics',3200,2,6400),
(119,'Arun Menon','Chennai','Cardiology',6000,1,6000),
(120,'Riya Shah','Ahmedabad','Dermatology',2000,2,4000),
(121,'Kunal Mehta','Lucknow','Orthopedics',3500,3,10500),
(122,'Sneha Joshi','Jaipur','Neurology',7000,2,14000),
(123,'Siddharth Gupta','Indore','Cardiology',5000,2,10000),
(124,'Anita Roy','Kolkata','Dermatology',1500,4,6000)
as patients_target(visit_id,patient_name,city,department,consultation_fee,tests_count,total_cost);
SELECT * FROM patients_target ORDER BY visit_id;
```

```python
%sql
CREATE OR REPLACE TEMP VIEW delta_updates as
select * from values
(115,'Rahul Verma','Hyderabad','Cardiology',6000,2,12000),
(117,'Vivek Nair','Kochi','Neurology',8000,1,8000),
(119,'Arun Menon','Chennai','Cardiology',6500,2,13000),
(121,'Kunal Mehta','Lucknow','Orthopedics',4000,3,12000),
(123,'Siddharth Gupta','Indore','Cardiology',5500,2,11000)
AS delta_updates(visit_id,patient_name,city,department,consultation_fee,tests_count,total_cost);
    
```

```python
%sql
MERGE INTO patients_target t
USING delta_updates u
ON t.visit_id = u.visit_id
WHEN MATCHED THEN 
UPDATE SET 
t.patient_name = u.patient_name,
t.city = u.city,
t.department = u.department,
t.consultation_fee = u.consultation_fee,
t.tests_count = u.tests_count,
t.total_cost = u.total_cost
WHEN NOT MATCHED THEN INSERT *;
```
