```python
data = [
(101,"Arjun Reddy","Hyderabad","Cardiology",5000),
(102,"Sneha Kapoor","Delhi","Orthopedics",3000),
(103,"Rahul Sharma","Mumbai","Dermatology",1500),
(104,"Priya Nair","Bangalore","Cardiology",5000),
(105,"Vikram Singh","Chennai","Neurology",7000)
]
columns = ["visit_id","patient_name","city","department","consultation_fees"]
df = spark.createDataFrame(data, columns)
display(df)
```

```python
df.write.mode("overwrite").parquet("/tmp/patient_parquet")
```

```python
parquet_df=spark.read.parquet("/tmp/patient_parquet")
display(parquet_df)
```

```python
parquet_df.printSchema()
```

```python
spark.read.parquet("/tmp/patient_parquet").select("patient_name","city").show()
```

```python
spark.read.parquet("/tmp/patient_parquet").filter("consultation_fees > 3000").show()
```

```python
df.write.mode("overwrite").partitionBy("city").parquet("/tmp/patient_parquet_partitioned")
```

```python
spark.read.parquet("/tmp/patient_parquet_partitioned").show()
```

```python
spark.read.parquet("/tmp/patient_parquet_partitioned").filter("city = 'Hyderabad'").show()
```
