**Section 6 — Joins**

```sql
--31
SELECT e.*,p.budget
FROM employees e 
JOIN projects p 
ON e.department=p.department;
```

```sql
--32
SELECT e.emp_name,e.department,p.project_name
FROM employees e 
JOIN projects p 
ON e.department=p.department;
```

```sql
--33
SELECT department,SUM(budget) as total_budget
FROM projects
GROUP BY department;
```

```sql
--34
SELECT DISTINCT e.*
FROM employees e 
JOIN projects p 
ON e.department=p.department
ORDER BY emp_id;
```

```sql
--35
SELECT p.project_name,COUNT(e.emp_id) AS emp_count
FROM employees e 
JOIN projects p 
ON e.department=p.department
GROUP BY p.project_name;
```

```sql
--39
SELECT *
FROM employees e
WHERE salary > (
  SELECT AVG(salary) 
  FROM employees
  WHERE department=e.department
);
```

**Final Mini Capstone**

```sql
CREATE TABLE employee_salary_analysis AS
SELECT emp_name,city,department,salary,experience,
IF(salary>=95000,'High',
  IF(salary>=80000,'Medium','Low')
) AS salary_category,
AVG(salary) OVER (PARTITION BY department) AS department_avg_salary
FROM employees;
```
