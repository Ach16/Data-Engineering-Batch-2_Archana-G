PART 1 — DataFrame Creation


```python
#ex1
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("Dataframe exercises").getOrCreate()
print("Spark session created")
```

```python
#ex2
emp_df=spark.read.csv("employees.csv",header=True,inferSchema=True)
#ex3
dept_df=spark.read.csv("departments.csv",header=True,inferSchema=True)
#ex4
sales_df=spark.read.csv("sales.csv",header=True,inferSchema=True)
#ex5
emp_nested_df=spark.read.option("multiline","true").json("employees_nested.json")
#ex6
emp_df.printSchema()
dept_df.printSchema()
sales_df.printSchema()
emp_nested_df.printSchema()
```

PART 3 — Selecting and Renaming

```python
#ex11
emp_df.select("name","salary").show()
```

```python
#ex12
emp_df=emp_df.withColumnRenamed("salary","emp_salary")
emp_df.show()
```

```python
#ex13
from pyspark.sql.functions import col
emp_df.filter(col("joining_date")>"2023-01-01").show()
```

PART 4 — Filtering Data

```python
#ex14
emp_df.filter(emp_df.emp_salary>65000).show()
```

```python
#ex15
emp_df.filter(emp_df.department=="IT").show()
```

```python
#ex16
emp_df.filter((emp_df.department=="IT")&(emp_df.emp_salary>70000)).show()
```

PART 5 — Dropping and Sorting

```python
#ex17
emp_df=emp_df.drop("joining_date")
emp_df.show()
```

```python
#ex18
emp_df.orderBy(emp_df.emp_salary).show()
```

```python
#ex19
emp_df.orderBy(emp_df.emp_salary.desc()).show()
```

```python
#ex20
emp_df.orderBy(emp_df.emp_salary.desc()).show(3)
```

PART 6 — Aggregations

```python
from pyspark.sql.functions import sum,avg,max,min
emp_df.select(
    sum("emp_salary").alias("total_salary"),  #ex21
    avg("emp_salary").alias("average_salary"),#ex22
    max("emp_salary").alias("highest_salary"),#ex23
    min("emp_salary").alias("lowest_salary")  #ex24
).show()
```

PART 7 — GroupBy

```python
#ex25
emp_df.groupBy("department").count().show()
```

```python
#ex26
emp_df.groupBy("department").avg("emp_salary").show()
```

```python
#ex27
emp_df.groupBy("department").sum("emp_salary").show()
```

PART 8 — Joining DataFrames

```python
#ex28
emp_df.join(dept_df,on="department",how="inner").show()
```

```python
#ex29
emp_df.join(dept_df,on="department",how="left").show()
```

```python
#ex30
emp_df.join(dept_df,on="department",how="inner").select("name","location").show()
```

PART 9 — Updating Data

```python
#ex31
emp_df=emp_df.withColumn("bonus",emp_df.emp_salary*0.1)
emp_df.show()
```

```python
#ex32
from pyspark.sql.functions import lit
emp_df=emp_df.withColumn("company",lit("BotCampus"))
emp_df.show()
```

PART 10 — Handling Null Values

```python
#ex33
emp_df.filter(col("department").isNull()).show()
```

```python
#ex34
emp_df.na.fill({"department":"Unknown"}).show()
```

```python
#ex35
emp_df.na.drop().show()
```

PART 11 — Built-in Functions

```python
from pyspark.sql.functions import upper,length
emp_df.select(
    "name",
    upper("name").alias("upper_name"), #ex36
    length("name").alias("name_length")#ex37
).show()
```

```python
#ex38
emp_df=emp_df.withColumn("salary_in_lakhs",col("emp_salary")/100000)
emp_df.show()
```
