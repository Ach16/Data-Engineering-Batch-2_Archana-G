```python
#86
spark.sql("""
SELECT
c.customer_id,
c.name,
c.city,
a.balance,
RANK() OVER (PARTITION BY c.city ORDER BY a.balance DESC) AS cust_rank
FROM customers c
JOIN accounts a
ON c.customer_id=a.customer_id
""").show()
```

```python
#87
spark.sql("""
SELECT *
FROM(
  SELECT
  c.customer_id,
  c.name,
  c.city,
  a.balance,
  ROW_NUMBER() OVER (PARTITION BY c.city ORDER BY a.balance DESC) AS row_num
  FROM customers c
  JOIN accounts a
  ON c.customer_id=a.customer_id
)t
WHERE row_num=1
""").show()
```

```python
#88
spark.sql("""
SELECT *,
DENSE_RANK() OVER (PARTITION BY txn_type ORDER BY amount DESC) AS txn_dense_rank
FROM transactions
""").show()
```

```python
#89
spark.sql("""
SELECT *
FROM(
  SELECT *,
ROW_NUMBER() OVER (PARTITION BY account_id ORDER BY amount DESC) AS row_num
FROM transactions
)t
WHERE row_num<=2
""").show()
```

```python
#90
spark.sql("""
SELECT *,
SUM(amount) OVER (PARTITION BY account_id ORDER BY txn_date,txn_id) AS running_total
FROM transactions
""").show()
```

```python
#91
spark.sql("""
SELECT
branch,
total_balance,
RANK() OVER (ORDER BY total_balance DESC) AS branch_rank
FROM(
  SELECT branch,
  SUM(balance) as total_balance
  FROM accounts
  GROUP BY branch
)t
""").show()
```

```python
#92
spark.sql("""
SELECT
city,
total_balance,
RANK() OVER (ORDER BY total_balance DESC) AS city_rank
FROM (
SELECT
c.city,
SUM(a.balance) AS total_balance
FROM customers c
JOIN accounts a
ON c.customer_id=a.customer_id
GROUP BY c.city
)t
""").show()
```

```python
#93
spark.sql("""
SELECT
name,
total_amount,
RANK() OVER (ORDER BY total_amount DESC) AS cust_rank
FROM (
SELECT
c.name,
SUM(t.amount) AS total_amount
FROM customers c
JOIN accounts a
ON c.customer_id=a.customer_id
JOIN transactions t
ON a.account_id=t.account_id
GROUP BY c.name
)t
""").show()
```

```python
#94
spark.sql("""
SELECT *
FROM (
SELECT
c.city,
c.name,
SUM(t.amount) AS total_amount,
ROW_NUMBER() OVER (PARTITION BY c.city ORDER BY SUM(t.amount) DESC) AS row_num
FROM customers c
JOIN accounts a
ON c.customer_id=a.customer_id
JOIN transactions t
ON a.account_id=t.account_id
GROUP BY c.city,c.name
) t
WHERE row_num=1
""").show()
```

```python
#95
spark.sql("""
SELECT *
FROM (
SELECT
b.region,
c.name,
a.balance,
ROW_NUMBER() OVER (PARTITION BY b.region ORDER BY a.balance DESC) AS row_num
FROM customers c
JOIN accounts a
ON c.customer_id=a.customer_id
JOIN branches b
ON a.branch=b.branch
)t
WHERE row_num=1
""").show()
```
