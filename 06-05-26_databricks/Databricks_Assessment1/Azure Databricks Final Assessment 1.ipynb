```python
%sql
-- 52
SELECT *,
RANK() OVER (PARTITION BY specialization ORDER BY total_revenue DESC) AS rank
FROM(
    SELECT specialization,doctor_name,SUM(bill_amount) AS total_revenue
    FROM final_data
    GROUP BY specialization,doctor_name
)t;
```

```python
%sql
-- 53
SELECT *
FROM (
    SELECT city,patient_name,SUM(bill_amount) AS total_bill,
    ROW_NUMBER() OVER (PARTITION BY city ORDER BY SUM(bill_amount) DESC) AS row_num
    FROM final_data
    GROUP BY city,patient_name
)t
WHERE row_num=1;
```

**Part 8 — Merge / Upsert**

```python
daily_visits_data = [
(26,1002,201,"2024-03-14","Completed",2),
(27,1005,210,"2024-03-14","Completed",1),
(11,1011,205,"2024-03-06","Completed",3),
(20,1020,206,"2024-03-10","Completed",2),
(28,1018,203,"2024-03-15","Pending",2)
]
daily_visits_columns = [
"visit_id","patient_id","doctor_id","visit_date","visit_status","tests_count"
]
daily_visits_df=spark.createDataFrame(daily_visits_data, daily_visits_columns)
```

```python
%sql
-- 71
CREATE OR REPLACE TABLE target_visits
USING DELTA
AS
SELECT *
FROM visits;
```

```python
%sql
-- 72
SELECT * FROM target_visits;
```

```python
#73
daily_visits_df.createOrReplaceTempView("daily_visits")
```

```python
%sql
-- 74, 75, 76
MERGE INTO target_visits target
USING daily_visits source
ON target.visit_id=source.visit_id

WHEN MATCHED THEN
UPDATE SET
target.patient_id=source.patient_id,
target.doctor_id=source.doctor_id,
target.visit_date=source.visit_date,
target.visit_status=source.visit_status,
target.tests_count=source.tests_count

WHEN NOT MATCHED THEN
INSERT (visit_id,patient_id,doctor_id,visit_date,visit_status,tests_count) VALUES 
(source.visit_id,source.patient_id,source.doctor_id,source.visit_date,source.visit_status,source.tests_count);
```

```python
%sql
-- 77
SELECT t.*
FROM target_visits t
INNER JOIN daily_visits d
ON t.visit_id=d.visit_id
WHERE t.visit_id IN (
SELECT visit_id
FROM visits
);
```

```python
%sql
-- 78
SELECT t.*
FROM target_visits t
LEFT JOIN visits v
ON t.visit_id=v.visit_id
WHERE v.visit_id IS NULL;
```

```python
%sql
-- 79
DESCRIBE HISTORY target_visits;
```

```python
#80
#  MERGE is used for incremental loads because it supports UPDATE and INSERT in a single transaction. Existing records are updated and new records are inserted efficiently.
```
