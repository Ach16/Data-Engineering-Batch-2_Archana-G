Dataset 1 — TXT (Website Login Logs)

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
spark=SparkSession.builder.appName("Pyspark challenge").getOrCreate()
print("Spark session created!")
```

```python
from google.colab import files
uploaded=files.upload()
```

```python
#1
logins_df=spark.read.text("logins.txt")
logins_df=logins_df.withColumnRenamed("value","name")
logins_df.show()
```

```python
#2
print("Total login events:",logins_df.count())
```

```python
#3
print("Unique users:")
logins_df.distinct().show()
```

```python
#4
print("Login counts of each user:")
login_count=logins_df.groupBy("name").count()
login_count.show()
```

```python
#5
print("Top 3 most active users:")
login_count.orderBy(col("count").desc()).show(3)
```

```python
#6
print("User who logged in more than 4 times:")
login_count.filter(col("count")>4).show()
```

```python
#7
login_dict={row['name']:row['count'] for row in login_count.collect()}
print(login_dict)
```

Dataset 2 — CSV (Employees Dataset)

```python
#1
emp_df=spark.read.csv("employees.csv",header=True,inferSchema=True)
emp_df.show()
```

```python
#2
print("Total employees count:",emp_df.count())
```

```python
#3
print("Employees working in IT dept:")
emp_df.filter(emp_df.department=="IT").show()
```

```python
#4
print("Employees with salary >75000:")
emp_df.filter(emp_df.salary>75000).show()
```

```python
#5,6,7
emp_df.select(avg("salary").alias("Average salary"),
              max("salary").alias("Highest salary"),
              min("salary").alias("Lowest salary")).show()
```

```python
#8
print("Employees per department:")
emp_df.groupBy("department").count().show()
```

```python
#9
print("Average salary per department:")
emp_df.groupBy("department").agg(avg("salary").alias("Average salary")).show()
```

```python
#10
print("Total employees per city:")
emp_df.groupBy("city").count().show()
```

```python
#11
print("Top 5 highest salaries:")
emp_df.orderBy(col("salary").desc()).show(5)
```

```python
#12
print("Employees working in Hyderabad with >70k salary:")
emp_df.filter((emp_df.city=="Hyderabad")&(emp_df.salary>70000)).show()
```

Dataset 3 — CSV (Sales Dataset)

```python
sales_df=spark.read.csv("sales.csv",header=True,inferSchema=True)
sales_df.show()
```

```python
#1
print("Revenue per sale: ")
sales_df=sales_df.withColumn("revenue",sales_df.price*sales_df.quantity)
sales_df.show()
```

```python
#2
print("Total revenue:")
sales_df.select(sum("revenue").alias("Total revenue")).show()
```

```python
#3
print("Revenue per product: ")
sales_df.groupBy("product").agg(sum("revenue").alias("Total revenue")).show()
```

```python
#4
print("Total quantity sold per product: ")
sold=sales_df.groupBy("product").agg(sum("quantity").alias("Total quantity"))
sold.show()
```

```python
#5
print("Best selling product: ")
sold.orderBy(col("Total quantity").desc()).show(1)
```

```python
#6
print("Employee generating highest revenue:")
sales_df.groupBy("emp_id").agg(sum("revenue").alias("Total revenue")).orderBy(col("Total revenue").desc()).show(1)
```

```python
#7
print("Average sale value:")
sales_df.select(avg("revenue").alias("Average sale value")).show()
```

```python
#8
print("Products with revenue above 100000:")
sales_df.groupBy("product").agg(sum("revenue").alias("revenue")).filter(col("revenue")>100000).show()
```

Dataset 4 — JSON (Orders Dataset)

```python
#1
orders_df=spark.read.option("multiline","true").json("orders.json")
orders_df=orders_df.select(explode("orders").alias("order")).select("order.*")
```

```python
#2
orders_df.show()
```

```python
#3
print("Total orders:",orders_df.count())
```

```python
#4
print("Total sales amount:")
orders_df.select(sum("amount")).show()
```

```python
#5
print("Total spending per customer:")
amt_spent=orders_df.groupBy("customer").agg(sum("amount").alias("Total spending"))
amt_spent.show()
```

```python
#6
print("Highest spending customer")
amt_spent.orderBy(col("Total spending").desc()).show(1)
```

```python
#7
print("Total sales per product")
orders_df.groupBy("product").agg(sum("amount").alias("Total sales")).show()
```

```python
#8
print("Customers from Hyderabad")
orders_df.filter(col("city")=="Hyderabad").show()
```

```python
#9
print("Orders with amount greater than 10,000:")
orders_df.filter(col("amount")>10000).show()
```

```python
#10
print("Count of orders placed in each city:")
orders_df.groupBy("city").count().show()
```

Final Combined Challenge

```python
#1
print("After joing employees.csv and sales.csv using emp_id:")
joined=emp_df.join(sales_df,on="emp_id",how="inner")
joined.show()
```

```python
#2
print("Total revenue generated by each employee:")
emp_rev=joined.groupBy("name").agg(sum("revenue").alias("Total revenue"))
emp_rev.show()
```

```python
#3
print("Top 5 employees by sales")
emp_rev.orderBy(col("Total revenue").desc()).show(5)
```

```python
#4
print("Department generating highest revenue")
joined.groupBy("department").agg(sum("revenue").alias("revenue")).orderBy(desc("revenue")).show(1)
```

```python
#5
emp_rev.write.csv("final_sales_report.csv",header=True,mode="overwrite")
print("Saved")
```
